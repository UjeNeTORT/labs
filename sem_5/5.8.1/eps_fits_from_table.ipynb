```python
# Аппроксимация ε(T) для вольфрама: обе зависимости — прямые линии
import numpy as np
import matplotlib.pyplot as plt

# Табличные данные
T = np.array([800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000], float)
eps_lambda_T = np.array([0.460, 0.458, 0.456, 0.454, 0.452, 0.450, 0.448, 0.446, 0.443, 0.441, 0.439, 0.437, 0.435], float)
eps_T = np.array([0.067, 0.081, 0.105, 0.119, 0.133, 0.144, 0.164, 0.179, 0.195, 0.209, 0.223, 0.236, 0.249], float)

def fit_line(x, y):
    A = np.vstack([x, np.ones_like(x)]).T
    a, b = np.linalg.lstsq(A, y, rcond=None)[0]
    return a, b

# --- 1. ε_{λ,T}(T)
a1, b1 = fit_line(T, eps_lambda_T)
x = np.linspace(T.min(), T.max(), 300)
plt.figure(figsize=(7,5), dpi=120)
plt.scatter(T, eps_lambda_T, label="данные", s=25)
plt.plot(x, a1*x + b1, label=f"ελ,T = {a1:.3e}·T + {b1:.3f}")
plt.xlabel("T, K")
plt.ylabel(r"$\varepsilon_{\lambda,T}$  (λ = 6500 Å)")
plt.title(r"Аппроксимация $\varepsilon_{\lambda,T}(T)$ линейной функцией")
plt.legend()
plt.grid(True)
plt.show()

print(f"ε_{'{λ,T}'}: a = {a1:.6e}, b = {b1:.6f}")

# --- 2. ε_T(T)
a2, b2 = fit_line(T, eps_T)
plt.figure(figsize=(7,5), dpi=120)
plt.scatter(T, eps_T, label="данные", s=25)
plt.plot(x, a2*x + b2, label=f"εT = {a2:.3e}·T + {b2:.3f}", color='orange')
plt.xlabel("T, K")
plt.ylabel(r"$\varepsilon_T$")
plt.title(r"Аппроксимация $\varepsilon_T(T)$ линейной функцией")
plt.legend()
plt.grid(True)
plt.show()

print(f"ε_T: a = {a2:.6e}, b = {b2:.6f}")
```
