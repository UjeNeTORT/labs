# сцинтилляционная гамма-спектрометрия

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
def load_spectrum(path, cols=(0,1)):
    data = pd.read_csv(path, header=None, usecols=list(cols))
    ch = data.iloc[:,0].to_numpy()
    counts = data.iloc[:,1].to_numpy()
    return ch, counts
```

```python
def plot_spectrum(ch, counts, xlim=None, ylim=None):
    plt.figure()
    plt.plot(ch, counts)
    if xlim is not None:
        plt.xlim(*xlim)
    if ylim is not None:
        plt.ylim(*ylim)
    plt.xlabel('Номер канала')
    plt.ylabel('Счёт, отн. ед.')
    plt.show()
```
